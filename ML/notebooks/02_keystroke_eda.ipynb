```python
plt.figure(figsize=(12,8))
sns.heatmap(keystroke.corr(numeric_only=True), cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
keystroke.hist(figsize=(15,10))
plt.tight_layout()
plt.show()
```

```python
numeric_cols = keystroke.select_dtypes(include=np.number).columns

for col in numeric_cols:
    plt.figure(figsize=(6,3))
    sns.boxplot(x=keystroke[col])
    plt.title(col)
    plt.show()
```

```python
sns.countplot(x=keystroke["UD"])
plt.title("Target Distribution")
plt.show()
```
