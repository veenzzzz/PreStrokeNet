```python
import pandas as pd
import numpy as np
```

```python
stroke = pd.read_csv(
    r"C:\Users\navee\PreStrokeNet\Datasets\raw\Stroke\healthcare-dataset-stroke-data.csv"
)

keystroke = pd.read_csv(
    r"C:\Users\navee\PreStrokeNet\Datasets\raw\keystoke\KeyStrokeDistance.csv"
)
```

```python
stroke.isnull().sum()
keystroke.isnull().sum()
```

```python
stroke = stroke.drop_duplicates()
keystroke = keystroke.drop_duplicates()
```

```python
stroke["bmi"] = stroke["bmi"].fillna(stroke["bmi"].median())
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

stroke["gender"] = le.fit_transform(stroke["gender"])
stroke["ever_married"] = le.fit_transform(stroke["ever_married"])
stroke["work_type"] = le.fit_transform(stroke["work_type"])
stroke["Residence_type"] = le.fit_transform(stroke["Residence_type"])
stroke["smoking_status"] = le.fit_transform(stroke["smoking_status"])
```

```python
from sklearn.preprocessing import LabelEncoder

le_subject = LabelEncoder()
le_key = LabelEncoder()

keystroke["subject"] = le_subject.fit_transform(keystroke["subject"])
keystroke["key"] = le_key.fit_transform(keystroke["key"])
```
