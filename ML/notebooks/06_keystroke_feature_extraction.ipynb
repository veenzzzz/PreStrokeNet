```python
import pandas as pd
import numpy as np
```

```python
keystroke = pd.read_csv(
    r"C:\Users\navee\PreStrokeNet\Datasets\interim\keystroke_clean.csv"
)
```

```python
keystroke.head()
```

```python
features = keystroke.groupby("subject").agg({
    "H": ["mean", "std", "min", "max"],
    "UD": ["mean", "std", "min", "max"],
    "DD": ["mean", "std", "min", "max"]
})
```

```python
features.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaled = scaler.fit_transform(features)
```

```python
scaled_features = pd.DataFrame(
    scaled,
    columns=features.columns,
    index=features.index
)
```

```python
scaled_features.head()
```

```python
scaled_features.to_csv(
    r"C:\Users\navee\PreStrokeNet\Datasets\Processed\keystroke_features_final.csv"
)
```
