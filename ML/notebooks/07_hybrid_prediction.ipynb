```python
import pandas as pd
import numpy as np
import joblib
```

```python
stroke_model = joblib.load(
    r"C:\Users\navee\PreStrokeNet\ML\saved_models\stroke_model.pkl"
)
```

```python
stroke = pd.read_csv(
    r"C:\Users\navee\PreStrokeNet\Datasets\Processed\stroke_features.csv"
)
```

```python
keystroke = pd.read_csv(
    r"C:\Users\navee\PreStrokeNet\Datasets\Processed\keystroke_features_final.csv"
)
```

```python
stroke.head()
keystroke.head()
```

```python
X = stroke.drop("stroke", axis=1)

prediction = stroke_model.predict(X)

probability = stroke_model.predict_proba(X)
```

```python
stroke["clinical_prediction"] = prediction

stroke["clinical_probability"] = probability[:,1]
```

```python
stroke.head()
```

```python
keystroke["typing_score"] = (
    keystroke.mean(axis=1, numeric_only=True)
)
```

```python
keystroke.head()
```

```python
stroke.to_csv(
    r"C:\Users\navee\PreStrokeNet\Datasets\Processed\clinical_predictions.csv",
    index=False
)

keystroke.to_csv(
    r"C:\Users\navee\PreStrokeNet\Datasets\Processed\keystroke_scores.csv",
    index=False
)
```
